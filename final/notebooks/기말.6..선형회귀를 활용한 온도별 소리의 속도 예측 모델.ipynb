# 선형회귀를 활용한 온도별 소리의 속도 예측 모델

## 데이터 준비

```python
# 소리의 속도 데이터
x_data = [0.0, 1.0, 2.0, 3.0]           # 온도(°C)
y_data = [331.0, 331.6, 332.2, 332.8]   # 속도(m/s)
```

## 인공지능 모델 학습
w0와 w1 찾기

```python
# 관련 함수들 정의

# 모델: y = w0 + w1 * x
def predict2(w, x):
    return w[0] + w[1] * x

# 오차의 제곱
def errorSquared(y, y_pred):
    return (y - y_pred) * (y - y_pred)

# 손실 함수: 오차 제곱의 합
def loss(errors):
    return sum(errors)

# w0의 기울기
def gradw0(w, x, y) :
  return -2 * (y - (w[0] + x * w[1]))

# w1의 기울기
def gradw1(w, x, y) :
  return -2 * (y - (w[0] + x * w[1])) * x
```

```python
# 인공지능 모델 학습
w = [0.00, 0.00]
epochs = 500        # 절편(331)이 커서 학습을 충분히 오래해야 함
learning_rate = 0.01

for epoch in range(epochs):
    error = []
    grad_w0 = []
    grad_w1 = []
    for x, y in zip(x_data, y_data):
        y_pred = predict2(w, x)
        error.append(errorSquared(y, y_pred))
        grad_w0.append(gradw0(w, x, y))
        grad_w1.append(gradw1(w, x, y))
    l = loss(error)
    w[0] = w[0] - learning_rate * sum(grad_w0)
    w[1] = w[1] - learning_rate * sum(grad_w1)

    print("Epoch", epoch, "\tw_0 =", '%.2f' % w[0],
          "\tw_1 =", '%.2f' % w[1], "\tloss =", '%.2f' % l)
```

## 학습된 인공지능 모델을 활용한 예측 결과
학습된 인공지능 모델을 활용하여 원하는 온도의 결과를 예측해 보자


```python
temperature = float(input("소리의 속도를 구하고 싶은 온도 입력: "))
print(f"{epochs} Epochs로 학습된 모델: y = {w[0]:.2f} + {w[1]: .2f} * x")
print(f"{temperature}℃에서 소리의 속도 예측값:{predict2(w, temperature):.1f} m/s")
print("[과학지식 참고]20℃에서 소리의 속도 343m/s")
```
